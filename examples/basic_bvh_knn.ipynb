# K-Nearest Neighbours using a BVH

Use `BVH.knn` to build a small point-neighborhood feature update.

```python
import torch
import torchbvh as tb

torch.cuda.manual_seed(0)
# Source point cloud
B, N, D, C = 32, 1000, 3, 8  # N points, D spatial dimensions, C feature channels
M = 500 # M queries
k = 4 # number of neighbors to query

source_points = torch.randn(N, D, device="cuda")
source_features = torch.randn(N, C, device="cuda", requires_grad=True)
query_points = torch.randn(M, D, device="cuda") # where we want to query neighbors

# Build BVH over source points and query neighbors, destroy tree after use
with tb.BVH(source_points) as bvh:
    neighbor_idx, distances = bvh.knn(query_points, k=k)

# example transformation on neighbor features
neighbor_features = source_features[neighbor_idx]
print(neighbor_features.shape)  # (M, k, C)
```
